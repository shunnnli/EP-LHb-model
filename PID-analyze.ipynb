#### Plot d_kd_history

```python
d_kd_history = all_batch_extras["d_kd_history"]

sorted_d_kd_history = sorted(
        d_kd_history.items(),
        key=lambda kv: (kv[0][3], kv[0][2], kv[0][1], kv[0][0])  # (num_r, max_b, omit, kd)
    )

# Create figure and axes
fig, ax = plt.subplots(figsize=(10, 6))

# Anticipatory Licks
for (kd, omit, max_b, num_r), d_kd_sessions in sorted_d_kd_history:
    # Determine standard_length from the first session's length
    if len(d_kd_sessions) == 0:
        continue  # skip if no sessions
    standard_length = len(d_kd_sessions[0])
    fixed_sessions = [
        np.pad(s, (0, standard_length - len(s)), mode='constant') if len(s) < standard_length else s[:standard_length]
        for s in d_kd_sessions
    ]
    d_kd_sessions_array = np.stack(fixed_sessions)
    avg = np.mean(d_kd_sessions_array, axis=0)
    sem = np.std(d_kd_sessions_array, axis=0) / np.sqrt(d_kd_sessions_array.shape[0])

    x = np.arange(1, avg.shape[0] + 1)
    ax.plot(x, avg, label=f"kd={kd}, omit={omit}, max_b={max_b}, num_r={num_r}")
    ax.fill_between(x, avg - sem, avg + sem, alpha=0.3)  # standard error band

ax.set_ylabel("d * kd")
ax.set_xlabel("Trial Num")
ax.set_title("D Component Impact")
ax.legend(loc="upper left", bbox_to_anchor=(1.02, 1.0), borderaxespad=0, frameon=False)
```
